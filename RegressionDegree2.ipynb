```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.preprocessing  import PolynomialFeatures
from sklearn.metrics import r2_score
```

```python
X_train = np.array([[1, 2], 
                    [2, 3], 
                    [3, 4], 
                    [4, 2], 
                    [5, 5], 
                    [6, 3], 
                    [7, 6], 
                    [8, 4], 
                    [9, 7], 
                    [10, 5]])
```

```python
y_train = np.array([5, 12, 25, 20, 55, 45, 90, 80, 140, 120])
y_test  = np.array([10, 48, 110])
```

```python
X_test = np.array([[2, 2], 
                   [5, 4], 
                   [8, 6]])
```

```python
plt.scatter(X_train[:,0], X_train[:,1] ,color='blue')
plt.scatter(X_test[:,0], X_test[:,1] ,color='red')
plt.show()
```

```python
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
print(model.coef_)
```

```python
print(model.intercept_)
```

```python
y_pred = model.predict(X_test)
```

```python
##Comparing Metrics MSE,MAE,R2 Score with Simple Regression
```

```python
mean_absolute_error(y_test, y_pred)
```

```python
mse = mean_squared_error(y_test, y_pred)
print(mse)
```

```python
print(np.sqrt(mse))
```

```python
r2_score(y_test, y_pred)
```

```python
## Regression Degree2
```

```python
poly = PolynomialFeatures(degree=2, include_bias=False)
X_train_p = poly.fit_transform(X_train) 
X_test_p  = poly.transform(X_test)
```

```python
print(X_train_p)   # x1,x2 =>  x1, x2, x1^2, x1*x2, x2^2
```

```python
print(X_test_p)   # x1,x2 =>  x1, x2, x1^2, x1*x2, x2^2
```

```python
model_2 = LinearRegression()
model_2.fit(X_train_p, y_train)
```

```python
y_pred_2 = model_2.predict(X_test_p)
y_pred_2
```

```python
r2_score(y_test, y_pred_2)
```

```python
mse = mean_squared_error(y_test, y_pred_2)
print(np.sqrt(mse))
```

```python
mean_absolute_error(y_test, y_pred_2)
```
